#### Conditional statements:
Conditional statements are used when the statements have to print based on conditions
or
we use conditional statements when we have a multiple different statements as to print
types of conditional statements:
1) if (inside logics we use if condition)
2) if else
3) continuous if else/ if else ladder
4) Nested if else
#### syntax for if else:
if (condition):
	statements
else:
	statements

#### syntax for continuous if else:

if(cond1):
	statements cond1

else:
	if(cond2):
		statements cond2
	else:
	if(cond2):
		statements cond2
	......
else:
	default statements

or 

if(cond1):
	statements cond1
elif(cond2):
	statements cond2
elif(cond3):
	statements cond3
else:
default statements

#### syntax nested if:
if(cond1):
	if(cond2):
		pstatements of if
	else:
		statements of else
else:
	statements of else

```python
'''Scenario: An online store offers discounts based on the purchase amount:
10% discount for purchases between $100 and $500
20% discount for purchases above $500
No discount for purchases below $100
Task: Write a program that takes the purchase amount as input and calculates the discount and final amount to be paid.'''

pa=float(input("enter purchase amount:"))
if pa<0:
    print("purchase amount cannot be negative")
else:
    if pa>500:
        d=0.20
    elif pa>=100:
        d=0.10
    else:
        d=0.00
    discount=pa*d
    fa=pa-discount
    print(f"discount:{discount:.2f}\nfinal amount:{fa:.2f}")
```
