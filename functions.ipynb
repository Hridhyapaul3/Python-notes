```python
#prime number or not without return
def prime1():
    stv=int(input("enter a start value:"))
    spv=int(input("enter a stop value:"))
    for i in range(stv,spv+1,1):
        if i==1:
            continue
        for j in range(2,i,1):
            if i%j==0:
                break
        else:
            print(i)
            
```

```python
prime1()
```

```python
# printing prime numbers using return. we compulsory use listt and append values in it if we donot use list return only eturns only first prime number in the range
def prime2():
    listt=[]
    stv=int(input("enter a start value:"))
    spv=int(input("enter a stop value:"))
    for i in range(stv,spv+1,1):
        if i==1:
            continue
        for j in range(2,i,1):
            if i%j==0:
                break
        else:
            listt.append(i)
    return listt
```

```python
prime2()
```

```python
for i in prime2():
    print(i)
```
