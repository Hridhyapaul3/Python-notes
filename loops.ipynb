#### Loops
A loop is a programming concept used to execute a block of code repeatedly until certain condition is met or the sequence is completed.

#### Types of loops:
* for loop
     1) for loop with sequence
	syntax:
	for iterative variable in sequence variable:
		statements
     2) for loop with range
	
	initialization - start value
	condition - stop value
    inc/dec -step value
	for(i=1, i<=10, i++) - in C

	syntax:
	range(start value, stop value, step size):
	default start value is 0
	default stop value is upto given condition
	default step size is 1
	
	for iterative variable in range( start value, stop value, step size):
	     statements
* while loop
 Syntax:

* nested loop
  1) for loop inside for loop
  2) for loop inside while loop
  3) while loop inside while loop
  4) while loop inside for loop
